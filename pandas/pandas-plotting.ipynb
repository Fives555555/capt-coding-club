# Histogram

```python
import pandas as pd

df = pd.read_csv("./data/HistoricalQuotes_clean.csv")
df.head()
```

```python
df.Close.plot(
    kind='hist',
    title='Distribution of Closing Prices',
    xlabel='Closing Price',
    ylabel='Frequency'
)
```
